## 0.3 How exercises work

Every exercise has an id like `3.1` (lesson 3, exercise 1) and is checked by calling

```python
check("3.1", your_answer)
```

There are three kinds of exercise, and you will meet all three in every lesson:

| kind | what you pass to `check` | example |
|---|---|---|
| **conceptual** (multiple choice or a short written answer) | a letter, or a string | `check("3.3", "b")` |
| **maths** (a number you derive by hand or compute) | a number | `check("3.2", 0.6)` |
| **coding** (a function you write) | the function itself | `check("3.1", rw_update)` |

The checker gives **gradual feedback**:

1. A wrong answer gets a *diagnosis* of what probably went wrong (for example "your result has the opposite sign") and the **first hint**.
2. A second wrong attempt gets a second hint; a third gets the last one.
3. After the hints are used up, the **worked solution** is shown together with the mistakes students most often make.
4. A correct answer gets a short explanation of *why* it is correct.

You can also, at any time:

```python
hint("3.1")      # see the next hint without attempting
reveal("3.1")    # show the solution (try first!)
reset("3.1")     # forget your attempts
progress()       # table of what you have done so far
```

Your attempts are saved in a small file (`.rlcourse_progress.json`) next to the notebooks, so you can close a notebook and continue later.

**Honest advice.** The hints are designed so that you will usually get there without the solution. Reading a solution you have not struggled with teaches almost nothing — resist `reveal` until the checker offers it.
